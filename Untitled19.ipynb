```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import networkx as nx
import matplotlib.pyplot as plt
```

```python
import kagglehub
import glob
import os
path = kagglehub.dataset_download("emrebayirr/udemy-course-dataset-categories-ratings-and-trends")
print("Path to dataset files:", path)

csv_files = glob.glob(os.path.join(path, "*.csv"))
if not csv_files:
    raise FileNotFoundError("No CSV file found in the downloaded dataset directory.")

file_path = csv_files[0]
df = pd.read_csv(file_path)

print("\nDataset Shape:", df.shape)
print("Columns available:", df.columns.tolist())
print(df.head(2))
```

```python
import gradio as gr

def gradio_recommend_courses(user_input_interests):

    if not user_input_interests:
        return "Please enter your interests to get recommendations."

    recommendations = recommend_courses(user_input_interests, top_n=5)

    return recommendations.to_html(index=False)


iface = gr.Interface(
    fn=gradio_recommend_courses,
    inputs=gr.Textbox(lines=3, label="Enter your learning interests:", placeholder="e.g., Python programming, data analysis, web development"),
    outputs=gr.HTML(label="Recommended Courses"),
    title="Course Recommendation System",
    description="Enter your interests and get a list of recommended courses based on our dataset."
)

iface.launch()
```

```python
df['mix'] = df['category'] + ' ' + df['headline'] + ' ' + df['objectives'] + ' ' + df['instructional_level']

vectorizer = TfidfVectorizer(stop_words='english')
tfidf_matrix = vectorizer.fit_transform(df['mix'])

def recommend_courses(user_interests_text, top_n=3):

    user_vec = vectorizer.transform([user_interests_text])

    similarity_scores = cosine_similarity(user_vec, tfidf_matrix).flatten()

    top_indices = similarity_scores.argsort()[-top_n:][::-1]

    recommendations = []
    for idx in top_indices:
        recommendations.append({
            'Title': df.iloc[idx]['title'],
            'Category': df.iloc[idx]['category'],
            'Level': df.iloc[idx]['instructional_level'],
            'Rating': df.iloc[idx]['rating'],
            'Similarity Score': round(float(similarity_scores[idx]), 4)
        })

    return pd.DataFrame(recommendations)
```
